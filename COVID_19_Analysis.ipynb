# **Grouping the data**

## State-wise cases confirmed :

```python
data.groupby('State/UnionTerritory')['Confirmed'].sum()
```

```python
m= data.groupby('State/UnionTerritory')['Confirmed'].sum()
m.plot.bar(figsize=(15,5))
plt.show()
```

State-wise cases cured :

```python
data.groupby('State/UnionTerritory')['Cured'].sum()
```

```python
m = data.groupby('State/UnionTerritory')['Cured'].sum()
m.plot.bar(figsize=(15,5))
plt.show()
```

State-wise death cases :

```python
data.groupby('State/UnionTerritory')['Deaths'].sum()
```

```python
m = data.groupby('State/UnionTerritory')['Deaths'].sum()
m.plot.bar(figsize=(15,5))
plt.show()
```

Sorting Data(No.of confirmed cases) in descending order :

```python
g=data.groupby('State/UnionTerritory')['Confirmed'].sum().sort_values(ascending = False)
g
```

```python
g.plot.bar(figsize=(15,5))
plt.show()
```

Sorting data(No.of cured case) in descending order :

```python
g=data.groupby('State/UnionTerritory')['Cured'].sum().sort_values(ascending = False)
g
```

```python
g.plot.bar(figsize=(15,5))
plt.show()
```

Sorting data(No.of deaths) ins descending order :

```python
g= data.groupby('State/UnionTerritory')['Deaths'].sum().sort_values(ascending = False)
g
```

```python
g.plot.bar(figsize=(15,5))
plt.show()
```
